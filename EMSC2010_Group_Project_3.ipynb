# Regression Analysis

```python
#finding polynomial to fit data
data = pd.DataFrame({"y": df_pop_percent_change['Population']}) #define the y-variable by selecting the 'Population' column
data["x_scaled"] = (df_year_mean['Annual_Rainfall_Mean'] - np.mean(df_year_mean['Annual_Rainfall_Mean'])) / np.std(df_year_mean['Annual_Rainfall_Mean']) #scale the x values to have a mean of 0 and stdev of 1
data["x2_scaled"] = data["x_scaled"] ** 2 #used for quadratic regression
data["x3_scaled"] = data["x_scaled"] ** 3 #used for cubic regression
data["x4_scaled"] = data["x_scaled"] ** 4 #used for quartic regression
```

```python
# Drop rows with NaN values from the data DataFrame
data_cleaned = data.dropna()

# Fit a straight-line model (first-order polynomial)
model_linear = bmb.Model("y ~ x_scaled", data_cleaned)
idata_linear = model_linear.fit(idata_kwargs={"log_likelihood": True})

# Fit a quadratic model (second-order polynomial)
model_quad = bmb.Model("y ~ x_scaled + x2_scaled", data_cleaned)
idata_quad = model_quad.fit(idata_kwargs={"log_likelihood": True})

# Fit a cubic model (third-order polynomial)
model_cubic = bmb.Model("y ~ x_scaled + x2_scaled + x3_scaled", data_cleaned)
idata_cubic= model_cubic.fit(idata_kwargs={"log_likelihood": True})

# Fit a quartic model (fourth-order polynomial)
model_quart = bmb.Model("y ~ x_scaled + x2_scaled + x3_scaled + x4_scaled", data_cleaned)
idata_quart = model_quart.fit(idata_kwargs={"log_likelihood": True})

#(progressbar=False) REMEMBER to put these back in the above orders before saving to github please.
```

```python
# Compare polynomial models
comparison = az.compare({
    "linear": idata_linear,
    "quadratic": idata_quad,
    "cubic": idata_cubic,
    "quartic": idata_quart
})
print(comparison) #print the result
```

Multi-polynomial regression is shown to be less relevant to us than Linear, so we return to a simpler Linear regression model

```python
datatwo = pd.DataFrame({"x": df_year_mean['Annual_Rainfall_Mean'], "y": df_pop_percent_change['Population']}) #put the data into a dataframe with variable names "x" and "y".
datatwo_cleaned = datatwo.dropna()
model = bmb.Model("y ~ x", datatwo_cleaned) #setup the first-order polynomial model
idata = model.fit(draws=2000, tune=2000, chains=8, progressbar=False, idata_kwargs={"log_likelihood": True}) #sample the posteriors #tune makes the first 2000 and throws them out # chains makes 8 versions of this
# Good to have multiple chains here, run as many as you can to check outcomes.
```

```python
#model priors to check
model.plot_priors()
plt.tight_layout()
plt.show()
```

```python
#modeeling priors
az.plot_trace(idata);
plt.tight_layout() #this spaces the plots out so that they don't overlap
```

```python
#plot the regression model for annual rainfall and kangaroo population size

plt.figure(figsize=(10, 6)) # Standardized figure size
x_range = np.linspace(0, 3, 100) # Predict across a range of x values from 0 to 3, 100 times
new_data = pd.DataFrame({"x": x_range}) #dataframe with the new x-values
model.predict(idata, data=new_data, kind='response_params') #predict the distribution of regression lines at each x-value

# Make random draws from the posterior of the regression lines
y_mean_draws = idata.posterior["mu"].values

# Reshape to (total_draws, x_points)
y_mean_draws = y_mean_draws.reshape(-1, len(x_range))

# Compute posterior mean and HDI at each x point
posterior_mean = y_mean_draws.mean(axis=0)
hdi = az.hdi(y_mean_draws, hdi_prob=0.95)

# Plot the results
plt.plot(df_year_mean, df_pop_percent_change, 'ok', label="Data")
plt.plot(x_range, posterior_mean, color="C1", label="Posterior mean")
plt.fill_between(
    x_range,
    hdi[:, 0],
    hdi[:, 1],
    alpha=0.3,
    color="C1",
    label="95% HDI",
    edgecolor = None
)
plt.xlabel('Annual Average of Monthly Rainfall (mm)')
plt.ylabel('Eastern Grey Kangaroo Population Change (%)')
plt.title('Annual Rainfall: Linear Regression Model') # Added a title
plt.legend()
plt.minorticks_on()
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.show() # Ensured plot is displayed
```

# Summer Rainfall Analysis

```python
datasm = pd.DataFrame({"x": df_summer_mean['Summer_Rainfall_Mean'], "y": df_pop_percent_change['Population']}) #put the data into a dataframe with variable names "x" and "y".
datasm_cleaned = datasm.dropna()
modelsm = bmb.Model("y ~ x", datasm_cleaned) #setup the first-order polynomial model
idatasm = modelsm.fit(draws=2000, tune=2000, chains=8, idata_kwargs={"log_likelihood": True}) #sample the posteriors #tune makes the first 2000 and throws them out # chains makes 8 versions of this
# Good to have multiple chains here, run as many as you can to check outcomes.
#(, progressbar=False)
```

```python
#plot regression model for summer rainfall and kangaroo population size
plt.figure(figsize=(10, 6)) # Standardized figure size
x_range_sm = np.linspace(0, 6, 100) # Predict across a range of x values from 0 to 3, 100 times
new_data_sm = pd.DataFrame({"x": x_range_sm}) #dataframe with the new x-values
modelsm.predict(idatasm, data=new_data_sm, kind='response_params') #predict the distribution of regression lines at each x-value

# Make random draws from the posterior of the regression lines
y_mean_draws_sm = idatasm.posterior["mu"].values

# Reshape to (total_draws, x_points)
y_mean_draws_sm = y_mean_draws_sm.reshape(-1, len(x_range_sm))

# Compute posterior mean and HDI at each x point
posterior_mean_sm = y_mean_draws_sm.mean(axis=0)
hdi_sm = az.hdi(y_mean_draws_sm, hdi_prob=0.95)

# Plot the results
plt.plot(df_summer_mean, df_pop_percent_change, 'ok', label="Data")
plt.plot(x_range_sm, posterior_mean_sm, color="C1", label="Posterior mean")
plt.fill_between(
    x_range_sm,
    hdi_sm[:, 0],
    hdi_sm[:, 1],
    alpha=0.3,
    color="C1",
    label="95% HDI",
    edgecolor = None
)
plt.xlabel('Summer Season Average of Monthly Rainfall (mm)')
plt.ylabel('Eastern Grey Kangaroo Population Change (%)')
plt.title('Summer Rainfall: Linear Regression Model') # Added a title
plt.legend()
plt.minorticks_on()
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.show() # Ensured plot is displayed
```

```python
# Predict across a range of x values for the summer model
x_range_summer_pred = x_range_sm # Use the already defined x_range for summer
new_data_summer_pred = new_data_sm # Use the already defined new_data for summer

# Predict the distribution of population changes (response) at each x-value for the summer model
# This populates idatasm.posterior_predictive
modelsm.predict(idatasm, data=new_data_summer_pred, kind='response')

# Make random draws from the posterior predictive distribution of population change for summer
y_pps_draws_summer = idatasm.posterior_predictive["y"].values

# Reshape to (total_draws, x_points)
y_pps_draws_summer = y_pps_draws_summer.reshape(-1, len(x_range_summer_pred))

# Compute mean and HDI of the posterior predictive distribution for summer
posterior_mean_summer_pps = y_pps_draws_summer.mean(axis=0)
hdi_mean_summer = az.hdi(idatasm.posterior["mu"].values.reshape(-1, len(x_range_summer_pred)), hdi_prob=0.95)
hdi_pps_summer = az.hdi(y_pps_draws_summer, hdi_prob=0.95)

# Plot both posterior for regression lines and the observations for summer
plt.figure(figsize=(10, 6))
plt.plot(df_summer_mean, df_pop_percent_change, 'ok', label="Data")
plt.plot(x_range_summer_pred, posterior_mean_summer_pps, color="C1", label="Posterior predictive mean")
plt.fill_between(
    x_range_summer_pred,
    hdi_mean_summer[:, 0],
    hdi_mean_summer[:, 1],
    alpha=0.3,
    color="C2", # Use a different color for mean HDI
    label="95% HDI (mean)",
    edgecolor = None
)

plt.fill_between(
    x_range_summer_pred,
    hdi_pps_summer[:, 0],
    hdi_pps_summer[:, 1],
    alpha=0.2,
    color="C1",
    label="95% HDI (predictive)",
    edgecolor = None
)

plt.legend()
plt.minorticks_on()
plt.xlabel('Summer Rainfall Mean (mm)')
plt.ylabel('Eastern Grey Kangaroo Population Change (%)')
plt.title('Summer Rainfall: Posterior Predictive Distribution')
plt.show()
```

The above plot shows that we cannot definitively use rainfall as a predictive tool for kangaroo population - even with Summer being our most significant season. The Kangaroo population is clearly dependent on *multiple variables* including rainfall. It highlights that even when a primary hypothesis isn't confirmed, the process itself yields significant learning and guides future, more nuanced investigations into the drivers of population dynamics.
